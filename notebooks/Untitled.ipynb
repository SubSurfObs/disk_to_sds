```python
import glob, os
from obspy import read, Stream

#st = read("../inbox/BGB4/data/2025/09/11/04/*BGB4.ms")
#st = read("../inbox/BGB4/data/2025/09/11/04/2025-09-11 0406 58 BGB4.ms")
files = glob.glob("../inbox/BGB4/data/2025/09/11/05/*.ms")
#files = glob.glob("../inbox/BGB4/data/2025/10/01/00/*.ms")



st = Stream()
for f in files:
    if os.path.getsize(f) == 0:
        print(f"Skipping empty file: {f}")
        continue
    try:
        st += read(f)
    except Exception as e:
        print(f"Skipping unreadable file: {f} ({e})")
```
